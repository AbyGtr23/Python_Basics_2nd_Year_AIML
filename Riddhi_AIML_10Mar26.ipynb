```python

#the above list has the daily sales of all the dogs of a week. I want you to write
#  a python program to change the data to  appropriate data types and then print the following output
sales_data=["1800","1200","900","1150","2560","1890","2200"]
print("Day1 sale: Rs", sales_data[0])
print("Day2 sale: Rs", sales_data[1])
print("Day3 sale: Rs", sales_data[2])
print("Day4 sale: Rs", sales_data[3])
print("Day5 sale: Rs", sales_data[4])
print("Day6 sale: Rs", sales_data[5])
print("Day7 sale: Rs", sales_data[6])

total = int(sales_data[0]) + int(sales_data[1]) + int(sales_data[2]) + int(sales_data[3]) + int(sales_data[4]) + int(sales_data[5]) + int(sales_data[6])

print("Total sales for the week : Rs", total)

#same in for loop
print("\nFOR LOOP")
sales_data = ["1800","1200","900","1150","2560","1890","2200"]

# convert string data to integer
sales_int = [int(i) for i in sales_data]

for i, j in enumerate(sales_int, start=1):
    print(f"Day {i} sale: Rs {j}")

print(f"Total sales Rs {sum(sales_int)}")
```
