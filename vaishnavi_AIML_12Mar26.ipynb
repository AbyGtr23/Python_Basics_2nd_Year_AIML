```python
sales_data = ["1800", "1200", "900", "1150", "2560", "1890", "2200"]

sales_data = [int(i) for i in sales_data]

total = 0

for i in range(len(sales_data)):
    print("Day", i+1, "Sales : Rs.", sales_data[i])
    total = total + sales_data[i]

print("Total Sales of the week Rs.", total)
```
